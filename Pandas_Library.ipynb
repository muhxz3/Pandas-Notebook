**Pandas Library**
<br>


*   It is used to work with datasets using python
*   Pandas - Panel data or Python Data Analysis





```python
# importing pandas
import pandas as pd
```

It has two types of data
<br>

1.   Series - It is a single column or like a 1D array
2.   Data Frame - a table where data is stored in the form of rows and columns


**Creating Series**

```python
a = [1,2,3,4,5]
s = pd.Series(a)
print(s)
```

**Index Labels**

```python
s = pd.Series(a, index = ['a','b','c','d','e'])
print(s)
```

**DataFrame from a Dictionary**

```python
cars = {
    "name": ["BMW","Audi","MG","Benz"],
    "rating": [8,7,4,9]
}

s = pd.DataFrame(cars)
print(s)
```

**Series from a Dictionary**

```python
python_score = {"Muhaz": 24, "Riya": 30, "Arun": 32}
s = pd.Series(python_score)
print(s)
```

**Key Observation:** Use dictionary itself while creating a series to have desired index labels.

**Series Basic Operations**

```python
s = pd.Series(
    [10,20,30],
    index = ['a','b','c']
)

print(s*2)
```

```python
s2 = pd.Series(
    [1,2,3],
    index = ['b','c','d']
)

print(s + s2)
```

**DataFrame from a numpy arrays**

```python
import numpy as np

a = np.array([[1,2],[3,4],[5,6]])
df = pd.DataFrame(a, columns = ['DS','DA'])
print(df)
```

DataFrame -> numpy array

```python
print(df.to_numpy())
```

**Type() function**

```python
print(type(df))
```

**Key Parameters associated with read_csv()**

1. sep - tells the pandas what separates values.

```python
df = pd.read_csv("Day7_Student_Dataset.csv", sep = "\t")
df
```

2. header()

```python
df = pd.read_csv("Day7_Student_Dataset.csv", header = None)
df
```

3. names()

```python
df = pd.read_csv("Day7_Student_Dataset.csv", names = ['a','b','c','d','e','f','g'])
df
```

```python
df = pd.read_csv("Day7_Student_Dataset.csv", index_col = "Student_ID")
df
```

4. usecols()

```python
df = pd.read_csv("Day7_Student_Dataset.csv", usecols = ['Student_ID','Age'])
df
```
